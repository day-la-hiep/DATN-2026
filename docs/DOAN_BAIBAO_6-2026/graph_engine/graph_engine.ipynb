```python
import json
from collections import defaultdict

with open("../knowledge_base/knowledge_graph/relations_auto.json", "r") as f:
    relations = json.load(f)

RELATION_WEIGHTS = {
    "HAS_SYMPTOM": 3,
    "HAS_RISK": 1,
    "TREATED_BY": 0,
    "CONFUSED_WITH": -1,
    "MAY_PROGRESS_TO": 0
}


def graph_reasoning(user_features):

    scores = defaultdict(float)

    explanations = defaultdict(list)

    for edge in relations:

        disease = edge["s"]

        relation = edge["r"]

        target = edge["t"]

        confidence = edge["confidence"]

        if target in user_features:

            weight = (
                RELATION_WEIGHTS.get(
                    relation,
                    0
                )
                * confidence
            )

            scores[disease] += weight

            explanations[disease].append({
                "matched": target,
                "relation": relation,
                "score": weight
            })

    return scores, explanations

user_features = [
    "asymmetry",
    "color_variation",
    "uv_exposure"
]

scores, explanations = graph_reasoning(
    user_features
)

print(scores)
```
